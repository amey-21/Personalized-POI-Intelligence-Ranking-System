```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
pois = pd.read_csv("../data/pois.csv")
travelers = pd.read_csv("../data/travelers.csv")
interactions = pd.read_csv("../data/interactions.csv")

print("POIs:", pois.shape)
print("Travelers:", travelers.shape)
print("Interactions:", interactions.shape)
```

```python
print("\nPOI missing values")
print(pois.isnull().sum())

print("\nTraveler missing values")
print(travelers.isnull().sum())

print("\nInteraction missing values")
print(interactions.isnull().sum())
```

```python
print("Duplicate POIs:", pois["poi_id"].duplicated().sum())
print("Duplicate travelers:", travelers["traveler_id"].duplicated().sum())

print(
    "Invalid traveler IDs:",
    (~interactions["traveler_id"].isin(travelers["traveler_id"])).sum()
)

print(
    "Invalid POI IDs:",
    (~interactions["poi_id"].isin(pois["poi_id"])).sum()
)
```

```python
poi_interactions = (
    interactions
    .groupby("poi_id")
    .size()
    .sort_values(ascending=False)
)

print(poi_interactions.describe())
```

```python
traveler_interactions = (
    interactions
    .groupby("traveler_id")
    .size()
    .sort_values(ascending=False)
)

print(traveler_interactions.describe())
```

```python
interaction_summary = (
    interactions
    .groupby("interaction_type")["interaction_weight"]
    .agg(["count", "mean"])
)

interaction_summary
```

```python
interaction_with_poi = interactions.merge(
    pois[
        [
            "poi_id",
            "category",
            "popularity",
            "tourist_level",
        ]
    ],
    on="poi_id",
    how="left",
)

print(
    interaction_with_poi
    .groupby("category")
    .size()
    .sort_values(ascending=False)
)
```

```python
print(
    interaction_with_poi
    .groupby("category")["interaction_weight"]
    .mean()
    .sort_values(ascending=False)
)
```
